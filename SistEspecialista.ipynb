```python
dietas = [
    {"Título": "Dieta A", "Objetivo": "Perder peso", "Peso": "Acima", "Restrições": "Nenhuma"},
    {"Título": "Dieta B", "Objetivo": "Perder peso", "Peso": "Acima", "Restrições": "Vegano"},
    {"Título": "Dieta C", "Objetivo": "Ganhar peso", "Peso": "Abaixo", "Restrições": "Nenhuma"},
    {"Título": "Dieta D", "Objetivo": "Ganhar peso", "Peso": "Abaixo", "Restrições": "Vegano"},
]

regras = [
    {"Condição": {"Objetivo": "Perder peso", "Peso": "Acima", "Restrições": "Nenhuma"}, "Recomendação": ["Dieta A"]},
    {"Condição": {"Objetivo": "Perder peso", "Peso": "Acima", "Restrições": "Vegano"}, "Recomendação": ["Dieta B"]},
    {"Condição": {"Objetivo": "Ganhar peso", "Peso": "Abaixo", "Restrições": "Nenhuma"}, "Recomendação": ["Dieta C"]},
    {"Condição": {"Objetivo": "Ganhar peso", "Peso": "Abaixo", "Restrições": "Vegano"}, "Recomendação": ["Dieta D"]},
]
```

```python
def recomendar_dietas(preferencias):
    recomendacoes = []
    for regra in regras:
        condicao = regra["Condição"]
        if (preferencias["Objetivo"] == condicao["Objetivo"] and
                preferencias["Peso"] == condicao["Peso"] and
                    preferencias["Restrições"] == condicao["Restrições"]):
            recomendacoes.extend(regra["Recomendação"])
    return recomendacoes
```

```python
def main():
    objetivo = input("Qual seu objetivo? ")
    peso = input("Qual seu peso? ")
    restricoes = input("Tem alguma restrição? ")

    preferencias = {
        "Objetivo": objetivo,
        "Peso": peso,
        "Restrições": restricoes
    }

    dietas_recomendados = recomendar_dietas(preferencias)

    if dietas_recomendados:
        print("Dieta Recomendada:")
        for dieta in dietas_recomendados:
            print(f"- {dieta}")
    else:
        print("Nenhuma recomendação disponível para suas preferências.")

if __name__ == "__main__":
    main()
```
